```python
#1
def reverse_list(reverse_my_list):
    return reverse_my_list[::-1]
list = [10,20,30,40,50]
result = reverse_list(list)
print(result)
```

```python
#2
def bigest_in_list(bigest_on):
    strings = [item for item in bigest_on if isinstance(item, str)]
    numbers = [item for item in bigest_on if isinstance(item, (int, float))]

    longest_string = max(strings, key=len) if strings else None
    largest_number = max(numbers) if numbers else None

    if longest_string is None:
        return largest_number

    if largest_number is None:
        return longest_string

    if len(longest_string) >= largest_number:
        return longest_string

    return largest_number


bigest_on = ['coconut', 'mango', 'elephant', 1200, 500]

print(bigest_in_list(bigest_on))
```

```python
#3
import math
def oglidosi(x1,y1,x2,y2):
    distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    return distance
result = oglidosi(1, 2, 4, 6)
print(result)
```
